```python
# =========================
# 00 CREATE PROJECT STRUCTURE
# Serie A Forecast Project
# =========================

from google.colab import drive
import os

# Mount Google Drive
drive.mount("/content/drive")

# Main project path
PROJECT_PATH = "/content/drive/MyDrive/Serie_A_Forecast"

# Folders to create
folders = [
    "data/raw",
    "data/processed",
    "notebooks",
    "models",
    "outputs"
]

# Create main project folder and subfolders
for folder in folders:
    path = os.path.join(PROJECT_PATH, folder)
    os.makedirs(path, exist_ok=True)

# Create placeholder files so folders are visible
placeholder_files = [
    "data/raw/.gitkeep",
    "data/processed/.gitkeep",
    "notebooks/.gitkeep",
    "models/.gitkeep",
    "outputs/.gitkeep"
]

for file in placeholder_files:
    file_path = os.path.join(PROJECT_PATH, file)
    with open(file_path, "w") as f:
        f.write("")

print("Project structure created successfully!")
print(PROJECT_PATH)
```
